```python
import random

from llm_project.utils import (
    get_processed_path,
    load_jsonl_records,
)
```

```python
def choose_five_ids(language: str) -> list[str]:
    en_rows = load_jsonl_records(
        get_processed_path(
            language=language,
            split="train",
            variant="en",
        )
    )

    target_rows = load_jsonl_records(
        get_processed_path(
            language=language,
            split="train",
            variant=language,
        )
    )

    en_ids = {
        str(row["alignment_id"])
        for row in en_rows
    }

    target_ids = {
        str(row["alignment_id"])
        for row in target_rows
    }

    common_ids = sorted(
        en_ids & target_ids
    )

    rng = random.Random(42)

    return rng.sample(
        common_ids,
        5,
    )
```

```python
print("German:", choose_five_ids("de"))
print("Swedish:", choose_five_ids("sv"))
```
