## 🧠 LLM Prompt Design and Evaluation Course

Welcome! In this notebook, we will explore how to design effective prompts for reasoning, summarization, and creative writing tasks using Hugging Face LLMs. We will also evaluate prompt effectiveness across different models to see how prompt formulation can impact outputs.

### 1️⃣ Setup Hugging Face Transformers

We will use the `transformers` library to load and interact with pre-trained LLMs. Make sure you have the library installed, and you have your Hugging Face API token if using hosted models.

```python
# Install transformers if not installed
# !pip install transformers[sentencepiece] --quiet

from transformers import pipeline

# Example: Load a text-generation model
model_name = 'bigscience/bloom-560m'  # Can be swapped with other HF LLMs
generator = pipeline('text-generation', model=model_name)
```

### 2️⃣ Designing Prompts

We will design prompts for three tasks: reasoning, summarization, and creative writing. A good prompt should be clear, task-specific, and sometimes provide examples (few-shot prompting).

```python
# Reasoning Prompt
reasoning_prompt = """
You are a logical assistant. Solve this problem step by step:
If John has 3 apples and gives 2 to Mary, how many apples does John have left?
"""

# Summarization Prompt
summarization_prompt = """
Summarize the following text in one sentence:
""" + """
Artificial intelligence is transforming the world. It helps humans in healthcare, education, and even creative arts.
"""

# Creative Writing Prompt
creative_prompt = """
Write a short, imaginative story about a talking cat who travels to space.
"""
```

### 3️⃣ Generating Responses

We will feed the prompts to the model and inspect the outputs.

```python
print("### Reasoning Output")
print(generator(reasoning_prompt, max_length=100, do_sample=True)[0]['generated_text'])

print("\n### Summarization Output")
print(generator(summarization_prompt, max_length=50, do_sample=True)[0]['generated_text'])

print("\n### Creative Writing Output")
print(generator(creative_prompt, max_length=200, do_sample=True)[0]['generated_text'])
```

### 4️⃣ Evaluating Prompt Effectiveness

We can evaluate prompts using both **qualitative** and **quantitative** measures:

1. **Clarity**: Does the response match the intent?
2. **Completeness**: Is the task fully addressed?
3. **Creativity** (for creative prompts): Is it imaginative?
4. **Consistency across models**: Do different models produce similar high-quality results?

We will try a simple comparison across two models.

```python
# Load another model for comparison
model_name2 = 'google/flan-t5-small'
generator2 = pipeline('text-generation', model=model_name2)

# Evaluate reasoning prompt
print("### Reasoning with Model 2")
print(generator2(reasoning_prompt, max_length=100)[0]['generated_text'])

# Evaluate summarization prompt
print("\n### Summarization with Model 2")
print(generator2(summarization_prompt, max_length=50)[0]['generated_text'])

# Evaluate creative prompt
print("\n### Creative Writing with Model 2")
print(generator2(creative_prompt, max_length=200)[0]['generated_text'])
```

### 5️⃣ Tips for Prompt Design

- **Be explicit**: Describe the task clearly.
- **Provide examples**: Few-shot learning often improves reasoning.
- **Iterate**: Slight wording changes can dramatically affect outputs.
- **Task-specific tuning**: Creative prompts benefit from open-ended language, reasoning prompts from step-by-step instructions.

Prompt evaluation is part art, part science. Encouraging experimentation is key! 🎨🧪
